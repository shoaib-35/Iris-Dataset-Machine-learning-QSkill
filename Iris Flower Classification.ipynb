# _**Modeling - Simple Manual Model**_

```python
def single_feature_prediction(petal_length):
    if petal_length < 1.96:
        return 0
    elif petal_length < 4.95:
        return 1
    else:
        return 2
```

```python
x_train[:, 2]
```

```python
manual_y_prediction = np.array([single_feature_prediction(val) for val in x_train[:,2]])
```

```python
manual_model_accuracy = np.mean(manual_y_prediction == y_train)
```

```python
print(f"Manual Model Accruracy: {manual_model_accuracy * 100:.2f}% ")
```

# _**Checking the Misclassifying points**_

```python
cross_val_predict(model, x_train, y_train, cv=5)
```

```python
y_pred = cross_val_predict(model, x_train, y_train, cv=5)
```

```python
Correct_prediction_mask = y_pred == y_train
```

```python
not_correct = ~Correct_prediction_mask
```

```python
x_train[not_correct]
```

```python
df_predictions = df_train.copy()
```

```python
df_predictions["correct_predictions"] = Correct_prediction_mask
```

```python
df_predictions.head()
```

```python
df_predictions["predictions"] = y_pred
```

```python
df_predictions["predictions_label"] = df_predictions["predictions"].map({0: "setosa", 1: "versicolor", 2: "virginica"})
```

```python
df_predictions.head(5)
```

```python
sns.scatterplot(x="petal length (cm)", y="petal width (cm)", data=df_predictions, hue="predictions_label")
```

```python
sns.scatterplot(x="petal length (cm)", y="petal width (cm)", data=df_predictions, hue="target_names")
```

```python
def plot_incorrect_predictions(df_predictions, x_axis, y_axis):
    fig, axs = plt.subplots(2,2, figsize = (10,10))
    axs = axs.flatten()
    sns.scatterplot(x = x_axis, y=y_axis, data=df_predictions, hue="predictions_label", ax= axs[0])
    sns.scatterplot(x = x_axis, y=y_axis, data=df_predictions, hue="target_names", ax= axs[1])
    sns.scatterplot(x = x_axis, y=y_axis, data=df_predictions, hue="correct_predictions", ax= axs[2])
    axs[3].set_visible(False)
    plt.show()
```

```python
plot_incorrect_predictions(df_predictions, "petal length (cm)", "petal width (cm)" )
```
